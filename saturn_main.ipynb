```python
import pandas as pd
from transformers import AutoModelForSequenceClassification, AutoTokenizer, pipeline
```

```python
MODEL = "cardiffnlp/twitter-roberta-base-sentiment"
tokenizer = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForSequenceClassification.from_pretrained(MODEL)
```

```python
sentiment_pipeline = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer)
```

```python
csv_file = "/content/instagram_comments_sample.csv"  # Update path if needed
df = pd.read_csv(csv_file)
```

```python
def analyze_sentiment(comment):
    result = sentiment_pipeline(comment)
    return result[0]['label']  # Extracting the sentiment label
```

```python
df["Sentiment"] = df["Comment"].apply(analyze_sentiment)
```

```python
df.to_csv("instagram_comments_with_sentiment.csv", index=False)
```

```python

def calculate_sentiment_score(label):
    if label == 'LABEL_0':
        return -1
    elif label == 'LABEL_1':
        return 0
    elif label == 'LABEL_2':
        return 1
    else:
        return 0

df["Sentiment_Score"] = df["Sentiment"].apply(calculate_sentiment_score)
total_score = df["Sentiment_Score"].sum()
total_comments = len(df)
percentage = (total_score / total_comments) * 100

print(f"Total Sentiment Score: {total_score}")
print(f"Total Number of Comments: {total_comments}")
print(f"Percentage: {percentage:.2f}%")
```
